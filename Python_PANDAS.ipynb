## **Python Demo 2:** Subsetting, slicing, and indexing of data structures  

### **subsetting** -> certain element (1)


```python
s = Global_PS4_sales
s
```

```python
df = HxH_Popularity_Poll
df
```

```python
s[2016]
```

```python
#iloc - m row, n col in indexes (i - generic index)
df.iloc[[3],[2]]
```

```python
#loc - m row, column name 
df.loc[[2],['Nen']]
```

### **Slicing** -> getting a portion

---

```python
s #with 0 = 2014
```

```python
s[1:4] #range of the data you want to extract; [1 (inclusive), 4 (exclusive)]
```

```python
df
```

```python
df[1:4]
```

```python
df.loc[:, ['Nen Type']] #empty spaces before and after ":" means it will take the whole given table
```

```python
df.loc[0:3, ['Nen Type']] #second index is inclusive because it pertains to the exact position/location of the element
```

```python
df.loc[[0,2],['Character', 'Votes']] 
# [index_1, index_2] means you will exclusively isolate this part of the table; it is particular about the order YOU give
```

```python
df.loc[[4,0,2,1],['Character', 'Votes']]
```

### **Indexing** -> with given conditions
- getting a certain element but with given conditions (boolean operators)
---

```python
s
```

```python
s[s>60]
```

```python
df
```

```python
df.loc[(df['Votes']>1000)&(df['Votes']<5000)] # locating elements from dictionary
```

```python
df.loc[df['Nen Type']=='Transmuter'] # == equivalence
```

```python
df.loc[(df['Nen']=='Kanmuru'), ['Character']] #asks for who is the character under Nen with Kanmuru
```

```python
df.loc[(df['Nen']=='Kanmuru')]
```

```python
df.loc[df['Nen Type']=='Transmuter',['Character','Nen','Votes']]
```
