### **A. Word Rotation Problem**

---

```python
def rotate_word(text):
    new_word = text[1:] + text[0]
    return new_word
```

```python
text = ["python", "logic", "code", "A"]
print("'python' ---> " + rotate_word(text[0]))
print("'logic' ---> " + rotate_word(text[1]))
print("'code' ---> " + rotate_word(text[2]))
print("'A' ---> " + rotate_word(text[3]))
```

### **B. Username Builder Problem**

---

```python
def make_username(first_name, last_name):
    username = first_name.lower().replace(" ", "") + "." + last_name.lower().replace(" ", "")
    return username
```

```python
first_name = ["Ada", "Alan", "Ana Maria"]
last_name = ["Lovelace", "Turing", "De Leon"]
print(first_name[0], ",", last_name[0], "---> " + make_username(first_name[0], last_name[0]))
print(first_name[1], ",", last_name[1], "---> " + make_username(first_name[1], last_name[1]))
print(first_name[2], ",", last_name[2], "---> " + make_username(first_name[2], last_name[2]))
```

### **C. Bookend Swap Problem**

---

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last, *middle, first]
```

```python
items = ["1", "2", "3", "4", "5", "6"]
print("Original List:", items)
print("Swapped List:", swap_bookends(items))
```

```python
items = ["red", "green", "blue"]
print("Original List:", items)
print("Swapped List:", swap_bookends(items))
```

```python
items = ["8", "3"]
print("Original List:", items)
print("Swapped List:", swap_bookends(items))
```
