### **A. Word Rotation Problem**

---

```python
text = input("Enter a word: ")
```

```python
def rotate_word(text):
    new_word = text[1:] + text[0]
    return new_word

print("The new rotated word is: " + rotate_word(text))
```

### **B. Username Builder Problem**

---

```python
first_name = input("Enter First Name: ")
```

```python
last_name = input("Enter Last Name: ")
```

```python
def make_username(first_name, last_name):
    username = first_name.lower().replace(" ", "") + "." + last_name.lower().replace(" ", "")
    return username

print("Confirm Username: " + make_username(first_name, last_name))
    
```

### **C. Bookend Swap Problem**

---

```python
items = [input("Enter first element: "), input("Enter second element: ")]
confirm_list = input("[Add more elements? (y/n)]: ")

while confirm_list != "n":
    items.append(input("Continue element: "))
    items
    confirm_list = input("[Add more elements? (y/n)]: ")
    
print("Current List of Items: ", items)
```

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last, *middle, first]

print("Original List:", items)
print("Swapped List:", swap_bookends(items))
```
