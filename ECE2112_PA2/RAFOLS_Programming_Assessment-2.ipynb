### **A. Reproducible Normalization Problem**
---

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10,101,size=(5,5))
```

```python
print(X)
```

```python
float(X.mean()), float(X.std())
```

```python
print("Array X has a rounded mean of: ", np.round(X.mean(),0))
```

```python
print("Array X has a rounded standard deviation of: ", np.round(X.std(),1))
```

```python
X_normalized = (X - X.mean()) / X.std()
print(X_normalized)
```

```python
print("Array X has a rounded normalized mean of: ", np.round(X_normalized.mean(),0))
```

```python
print("Array X has a rounded normalized standard deviation of: ", np.round(X_normalized.std(),1))
```

```python
np.save("X_normalized.npy", np.ndarray)
```

### **B. Cubes Divisible by 4 Problem**
---

```python
C = np.arange(1, 101).reshape(10,10) ** 3
print(C)
```

```python
div_by_4 = C[C % 4 == 0]
print(div_by_4)
```

```python
print("The total number of cubed values that are divisible by 4: ", div_by_4.shape)
```

```python
np.save("div_by_4.npy", np.ndarray)
```

### **C. Above-Mean Squares Problem**
---

```python
S = np.arange(1, 37).reshape(6,6) ** 2
print(S)
```

```python
S_mean = S.mean()
print("Array S has a rounded mean of: ", np.round(S_mean,2))
```

```python
above_mean = S[S > S_mean]
print(above_mean)
```

```python
print("The total number of selected elements that are greater than the mean: ", above_mean.shape)
```

```python
np.save("above_mean.npy", np.ndarray)
```
