### **A. Reproducible Normalization Problem**
---

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10,101,size=(5,5))
```

```python
X
```

```python
X.mean(), X.std()
```

```python
print("Array X has a rounded mean of: ", np.round(X.mean(),0))
```

```python
print("Array X has a rounded standard deviation of: ", np.round(X.std(),1))
```

```python
X_normalized = (X - X.mean()) / X.std()
print(X_normalized)
```

```python
print("Array X has a rounded mean of: ", np.round(X_normalized.mean(),0))
```

```python
print("Array X has a rounded standard deviation of: ", np.round(X_normalized.std(),1))
```

```python
np.save("X_normalized.npy", np.ndarray)
```

### **B. Cubes Divisible by 4 Problem**
---

```python
C = np.arange(1, 101)
C.shape = (10,10)
print(C)
```

```python
C = np.power(C,3)
print(C)
```

```python
check_4 = (C % 4 == 0)
div_by_4 = C[check_4]
print(div_by_4)
```

```python
print("The total number of cubed values that are divisible by 4: ", div_by_4.shape)
```
