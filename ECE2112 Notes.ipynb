## Conditional Statement - simple IF statement

```python
#syntax: 
#if <expr>:
    #<statement>
```

```python
#Example:
a = int(input("Enter a number for a = : ")); b = int(input("Enter a number for b: "))
if b > a:
    print("b is greater than a")
```

```python
#Another syntax:
#if <expr>: <statement>
```

```python
#Example:
grade = int(input("Enter a grade for ECE2112 = : "))
if grade < 60:
    print("Im sorry")
    print("You failed")
else:
    print("You passed!")
```

Looping Statement
for loop - need of sequence (string, set, list, dictionary, tuple)

```python
for x in "banana":
    print(x)
```

```python
#break statement - stops the loop at a certain point
fruits = ["apple", "banana", "cherry", "mango"]
for x in fruits:
    print(x)
    if x =="mango":
        break
```

```python
#continue statement - stops at the current iteration then proceeds to the next
fruits = ["apple", "banana", "cherry", "mango"]
for x in fruits:
    if x =="banana":
        continue
    print(x)
```

## while statement

```python
i = 1
while i < 6:
    print(i)
    i += 1 #
```
